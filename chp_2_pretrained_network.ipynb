```python
import os
import torchvision.models as models

# 1. Set your custom directory path
os.environ['TORCH_HOME'] = '/media/tamal/New_HardDrive/Machine Learning, AI/Deep_learning_with_pytorch'

# 2. Instantiate model (it will download to / load from your custom path above)
vit = models.vit_b_16(weights=models.ViT_B_16_Weights.IMAGENET1K_V1)
```

```python
vit
```

```python
from torchvision import transforms

preprocess = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.224, 0.224, 0.225]
    )
])
```

```python
from PIL import Image

img = Image.open("dog.jpg")
# img.show()
```

```python
img_t = preprocess(img)
# img_t
```

```python
batch_t = torch.unsqueeze(img_t, 0)
# batch_t
```

```python
# to do inference, we need to put the network in eval mode
vit.eval()
```

```python
out = vit(batch_t)
out
```

```python
with open("imagenet_classes.txt") as f:
    labels = [line.strip() for line in f.readlines()]

labels
```

```python
_, index  = torch.max(out, 1)
```

```python
_, index
```

```python
percentage = torch.nn.functional.softmax(out, dim=1)[0] * 100
labels[index.item()], percentage[index.item()].item()
```

```python
labels[index.item()]
```

```python
# we also can find the second best value, third best and so on values. we can do this with sorting in descending oder

_, indices = torch.sort(out, descending=True)
[(labels[idx], percentage[idx].item()) for idx in indices[0][:5]]
```

### A pretrained network that describe scenes

```python
from PIL import Image
from transformers import BlipProcessor, BlipForConditionalGeneration

processor = BlipProcessor.from_pretrained("Salesforce/blip-image-captioning-large")

model = BlipForConditionalGeneration.from_pretrained("Salesforce/blip-image-captioning-large")
```

```python
# the helper funtion 

def annonate_image(image: Image) -> None:
    display(image)
    inputs = processor(image, return_tensors="pt")
    out = model.generate(**inputs)
    print(processor.decode(out[0], skip_special_tokens = True))
```

```python
annonate_image(Image.open("dog.jpg"))
```
