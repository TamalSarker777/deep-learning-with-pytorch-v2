### Named tensors

```python
img_t = torch.randn(3,5,5) # channels, rows, columns
weights  = torch.tensor([0.2126, 0.7152, 0.0722])

img_t, img_t.shape
```

```python
batch = torch.randn(2,3,5,5) # batch, channels, rows, columns
batch
```

```python
img_gray_native = img_t.mean(-3)
batch_gray_native = batch.mean(-3)

img_gray_native.shape, batch_gray_native.shape
```

```python
weights
```

```python
unsqueezed_weights = weights.unsqueeze(-1).unsqueeze(-1)
unsqueezed_weights
```

```python
img_weight = img_t * unsqueezed_weights
img_weight
```

```python
batch_weight = batch * unsqueezed_weights
batch_weight
```

```python
img_gray_weighted = img_weight.sum(-3)
batch_gray_weighted = batch_weight.sum(-3)

batch_weight.shape, batch.shape, unsqueezed_weights.shape
```

```python
# Named tensors allow aligning dimensions automatically by name using align_as
img_named = img_t.refine_names(..., 'channels', 'rows', 'columns')
batch_named = batch.refine_names(..., 'channels', 'rows', 'columns')
weights_named = torch.tensor([0.2126, 0.7152, 0.0722], names=('channels',))

# align_as adds dimensions of size 1 and permutes dimensions to match the target tensor
weights_aligned = weights_named.align_as(img_named)
print("weights_aligned shape:", weights_aligned.shape, "names:", weights_aligned.names)

# Now broadcasting works cleanly across named dimensions
gray_named = (img_named * weights_aligned).sum('channels')
print("gray_named shape:", gray_named.shape, "names:", gray_named.names)
```

## Tensors: Scenic views of storage
### Indexing into storage
multiple tensors can index the same storage even if they index the data differently. A storage is one dimentional array

```python
points = torch.tensor([[4,1],[5,3],[2,1]]).to(torch.float32)
print(points)
print("---------------------")
points.storage()
```

```python
# we can index into a storage manually

points_storage = points.storage()
print(points_storage[0])
points_storage[1]
```
