### Named tensors

```python
img_t = torch.randn(3,5,5) # channels, rows, columns
weights  = torch.tensor([0.2126, 0.7152, 0.0722])

img_t, img_t.shape
```

```python
batch = torch.randn(2,3,5,5) # batch, channels, rows, columns
batch
```

```python
img_gray_native = img_t.mean(-3)
batch_gray_native = batch.mean(-3)

img_gray_native.shape, batch_gray_native.shape
```

```python
weights
```

```python
unsqueezed_weights = weights.unsqueeze(-1).unsqueeze(-1)
unsqueezed_weights
```

```python
img_weight = img_t * unsqueezed_weights
img_weight
```

```python
batch_weight = batch * unsqueezed_weights
batch_weight
```

```python
img_gray_weighted = img_weight.sum(-3)
batch_gray_weighted = batch_weight.sum(-3)

batch_weight.shape, batch.shape, unsqueezed_weights.shape
```
